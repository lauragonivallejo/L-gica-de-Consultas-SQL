```python
# 1. Función que recibe una cadena y devuelve un diccionario con las frecuencias
# de cada letra. Los espacios no se cuentan.

def frecuencia_letras(texto):
    frecuencias = {}
    for letra in texto.lower():
        if letra != " ":
            frecuencias[letra] = frecuencias.get(letra, 0) + 1
    return frecuencias

print(frecuencia_letras("Hola Mundo"))
```

```python
# 2. Dada una lista de números, obtener una nueva lista con el doble de cada
# valor. Usa map()

def doblar(numeros):
    return list(map(lambda n: n * 2, numeros))

print(doblar([1, 2, 3, 4, 5]))
```

```python
# 3. Función que toma una lista de palabras y una palabra objetivo, y devuelve
# las palabras de la lista que contengan la palabra objetivo.

def buscar_palabras(palabras, objetivo):
    return [p for p in palabras if objetivo in p]

print(buscar_palabras(["gato", "gatos", "perro", "regato"], "gato"))
```

```python
# 4. Función que calcula la diferencia entre los valores de dos listas. Usa map()

def diferencia(lista1, lista2):
    return list(map(lambda a, b: a - b, lista1, lista2))

print(diferencia([10, 20, 30], [1, 2, 3]))
```

```python
# 5. Función que toma una lista de números y un valor opcional nota_aprobado
# (por defecto 5). Calcula la media y devuelve una tupla con la media y el
# estado ("aprobado" si la media >= nota_aprobado, "suspenso" si no).

def media_notas(notas, nota_aprobado=5):
    media = sum(notas) / len(notas)
    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"
    return (media, estado)

print(media_notas([6, 7, 8]))
print(media_notas([2, 3, 4]))
```

```python
# 6. Función que calcula el factorial de un número de manera recursiva.

def factorial(n):
    if n <= 1:          # caso base: aquí para la recursión
        return 1
    return n * factorial(n - 1)     # la función se llama a sí misma

print(factorial(5))
```

```python
# 9. Función que toma una lista de mascotas y devuelve una nueva lista
# excluyendo las prohibidas en España. Usa filter()

mascotas_prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]

def filtrar_mascotas(mascotas):
    return list(filter(lambda m: m not in mascotas_prohibidas, mascotas))

print(filtrar_mascotas(["Perro", "Tigre", "Gato", "Oso", "Canario"]))
```

```python
# 10. Función que calcula el promedio de una lista. Si la lista está vacía,
# lanza una excepción personalizada y maneja el error.

class ListaVaciaError(Exception):
    pass

def promedio(numeros):
    if len(numeros) == 0:
        raise ListaVaciaError("La lista está vacía")
    return sum(numeros) / len(numeros)

try:
    print(promedio([4, 8, 12]))
    print(promedio([]))
except ListaVaciaError as error:
    print("Error:", error)
```

```python
# 13. Función que para un conjunto de caracteres devuelve una lista de tuplas
# con cada letra en mayúscula y minúscula, sin repetir. Usa map()

def mayus_minus(caracteres):
    letras = sorted(set(caracteres.lower()))    # set() quita las repetidas
    return list(map(lambda l: (l.upper(), l), letras))

print(mayus_minus("Bootcamp"))
```

```python
# 17. Función que toma una lista de dígitos y devuelve el número correspondiente.
# Por ejemplo [5,7,2] -> 572. Usa reduce()

def digitos_a_numero(digitos):
    # en cada paso multiplicamos por 10 y sumamos el siguiente dígito:
    # 5 -> 5*10+7=57 -> 57*10+2=572
    return reduce(lambda acumulado, d: acumulado * 10 + d, digitos)

print(digitos_a_numero([5, 7, 2]))
```

```python
# 28. Función que busca y devuelve el primer elemento duplicado de una lista.

def primer_duplicado(lista):
    vistos = []
    for elemento in lista:
        if elemento in vistos:
            return elemento
        vistos.append(elemento)
    return None

print(primer_duplicado([1, 2, 3, 2, 5]))
print(primer_duplicado([1, 2, 3]))
```

```python
# 29. Función que convierte una variable en texto y enmascara todos los
# caracteres con '#', excepto los últimos cuatro.

def enmascarar(variable):
    texto = str(variable)
    if len(texto) <= 4:
        return texto
    return "#" * (len(texto) - 4) + texto[-4:]   # texto[-4:] son los 4 últimos

print(enmascarar(4556364607935616))
```

```python
# 32. Función que busca un nombre completo en una lista de empleados y devuelve
# su puesto, o un mensaje si la persona no trabaja aquí.

empleados = [
    {"nombre": "Alicia Perez", "puesto": "Data Analyst"},
    {"nombre": "Bob Martinez", "puesto": "Data Engineer"},
]

def buscar_puesto(nombre):
    for empleado in empleados:
        if empleado["nombre"] == nombre:
            return empleado["puesto"]
    return nombre + " no trabaja aqui"

print(buscar_puesto("Alicia Perez"))
print(buscar_puesto("Juan Nadie"))
```

```python
# 34. Clase Arbol con tronco y ramas como atributos y los métodos
# crecer_tronco, nueva_rama, crecer_ramas, quitar_rama e info_arbol.

class Arbol:
    def __init__(self):
        self.tronco = 1
        self.ramas = []

    def crecer_tronco(self):
        self.tronco += 1

    def nueva_rama(self):
        self.ramas.append(1)

    def crecer_ramas(self):
        self.ramas = [rama + 1 for rama in self.ramas]

    def quitar_rama(self, posicion):
        self.ramas.pop(posicion)

    def info_arbol(self):
        print("Longitud del tronco:", self.tronco)
        print("Numero de ramas:", len(self.ramas))
        print("Longitud de las ramas:", self.ramas)
```

```python
# 34. Caso de uso del árbol

arbol = Arbol()             # 1. crear el árbol
arbol.crecer_tronco()       # 2. crecer el tronco
arbol.nueva_rama()          # 3. nueva rama
arbol.crecer_ramas()        # 4. crecer las ramas
arbol.nueva_rama()          # 5. dos ramas más
arbol.nueva_rama()
arbol.quitar_rama(2)        # 6. quitar la rama de la posición 2
arbol.info_arbol()          # 7. información del árbol
```

```python
# 36. Clase UsuarioBanco con nombre, saldo y cuenta corriente, y los métodos
# retirar_dinero, transferir_dinero y agregar_dinero.

class UsuarioBanco:
    def __init__(self, nombre, saldo, cuenta_corriente):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    def retirar_dinero(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError("Saldo insuficiente")
        self.saldo -= cantidad
        print(self.nombre, "retira", cantidad, "- Saldo:", self.saldo)

    def transferir_dinero(self, origen, cantidad):
        if cantidad > origen.saldo:
            raise ValueError("Saldo insuficiente en " + origen.nombre)
        origen.saldo -= cantidad
        self.saldo += cantidad
        print("Transferencia de", cantidad, "de", origen.nombre, "a", self.nombre)

    def agregar_dinero(self, cantidad):
        self.saldo += cantidad
        print(self.nombre, "ingresa", cantidad, "- Saldo:", self.saldo)
```

```python
# 36. Caso de uso del banco
# Ojo: Bob se queda en 70 y se le piden 80, así que la transferencia falla.
# Por eso la metemos en un try/except.

alicia = UsuarioBanco("Alicia", 100, True)
bob = UsuarioBanco("Bob", 50, True)

bob.agregar_dinero(20)

try:
    alicia.transferir_dinero(bob, 80)
except ValueError as error:
    print("Error:", error)

alicia.retirar_dinero(50)
```

```python
# 37. Función procesar_texto que procesa un texto según la opción indicada:
# contar_palabras, reemplazar_palabras o eliminar_palabra.

def contar_palabras(texto):
    conteo = {}
    for palabra in texto.split():
        conteo[palabra] = conteo.get(palabra, 0) + 1
    return conteo

def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    return texto.replace(palabra_original, palabra_nueva)

def eliminar_palabra(texto, palabra):
    return " ".join([p for p in texto.split() if p != palabra])

def procesar_texto(texto, opcion, *args):   # *args = nº variable de argumentos
    if opcion == "contar":
        return contar_palabras(texto)
    elif opcion == "reemplazar":
        return reemplazar_palabras(texto, args[0], args[1])
    elif opcion == "eliminar":
        return eliminar_palabra(texto, args[0])
    else:
        return "Opcion no valida"

texto = "el gato duerme el gato come"
print(procesar_texto(texto, "contar"))
print(procesar_texto(texto, "reemplazar", "gato", "perro"))
print(procesar_texto(texto, "eliminar", "el"))
```

```python
# 40. Función que toma una figura ("rectangulo", "circulo" o "triangulo") y una
# tupla con los datos necesarios, y calcula el área.

def calcular_area(figura, datos):
    if figura == "rectangulo":
        return datos[0] * datos[1]
    elif figura == "circulo":
        return math.pi * datos[0] ** 2
    elif figura == "triangulo":
        return datos[0] * datos[1] / 2
    else:
        return "Figura no valida"

print(calcular_area("rectangulo", (4, 5)))
print(calcular_area("circulo", (3,)))       # ojo a la coma: es una tupla
print(calcular_area("triangulo", (6, 4)))
```

```python
# 41. Programa que calcula el precio final de una compra aplicando un cupón de
# descuento si el usuario lo tiene y es válido.

precio = float(input("Precio original del articulo: "))
tiene_cupon = input("¿Tienes cupon de descuento? (si/no): ")

if tiene_cupon == "si":
    descuento = float(input("Valor del cupon: "))
    if descuento > 0:
        precio_final = precio - descuento
        print("Descuento aplicado de", descuento)
    else:
        precio_final = precio
        print("El cupon no es valido")
else:
    precio_final = precio

print("Precio final:", precio_final)
```
